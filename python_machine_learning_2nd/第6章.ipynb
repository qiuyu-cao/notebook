## 6.2 开发模型

### 6.2.1 准备数据

#### 1. 向量化

神经网络的所有输入和目标通常都必须是浮点数张量。无论要处理什么数据，首先都必须将其转换为张量，这一步叫作**数据向量化**（data vectorization）

#### 2. 规范化

一般来说，将取值较大的数据输入到神经网络是不安全的。这么做可能导致较大的梯度更新，进而导致神经网络无法收敛。为了让神经网络的学习变得更容易，输入数据应具有以下特征：

- **取值较小**：通常情况下，大部分取值应该在 0 ~ 1 范围内。

- **同质性**：所有特征的取值应该大致在同一范围内。

两种常见的规范化方法：

- 对每个特征分别规范化，使其均值为 0。`x -= x.mean(axis=0)`

- 对每个特征分别规范化，使其标准差为 1。`x /= x.std(axis=0)`

#### 3. 处理缺失值

处理缺失值的方法：

- 舍掉这一条数据

- 如果使分类数据，可以再创建一个“缺失值”的分类

- 如果使数值特征，可以将缺失值替换为该特征的均值或中位值。避免使用 0，因为它可能会在特征形成的潜在空间造成不连续性。
